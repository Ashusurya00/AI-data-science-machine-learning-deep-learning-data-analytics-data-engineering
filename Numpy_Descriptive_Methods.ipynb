```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4])
arr
```

```python
arr2D = np.array([[1, 2], [3, 4]])
arr2D
```

#**Sum**


*   computes the sum of array elements

*   syntax : np.sum(arr,axis=None)



##**1D Array**

```python
arr
```

```python
np.sum(arr)
```

##**2D Array**

```python
arr2D
```

```python
# sum of all elements
np.sum(arr2D)
```

```python
# column-wise sum
np.sum(arr2D, axis=0)
```

```python
# row-wise sum
np.sum(arr2D, axis=1)
```
