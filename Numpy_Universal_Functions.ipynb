#**Universal Functions**

##**abs()**

*  Returns the absolute value of each element in the array.
*  syntax : np.abs(arr)



###**1D Array**

```python
arr = np.array([-1, -2, 3, -4])
arr
```

```python
np.abs(arr)
```

###**2D Array**

```python
arr = np.array([[-1, -2], [3, -4]])
arr
```

```python
np.abs(arr)
```

##**sqrt()**

*   Computes the square root of each element.

*   syntax : np.sqrt(arr)





###**1D Array**

```python
arr = np.array([4,9,16,25])
arr
```

```python
np.sqrt(arr)
```

###**2D Array**

```python
arr = np.array([[4,9], [16,25]])
arr
```

```python
np.sqrt(arr)
```

##**log()**

*   Computes the natural log (base e) of each element
*   syntax : np.log(arr)



###**1D Array**

```python
arr = np.array([1, np.e, np.e**2])
arr
```

```python
np.log(arr)
```

###**2D Array**

```python
arr = np.array([[1, np.e], [np.e**2, np.e**3]])
arr
```

```python
np.log(arr)
```

##**sign()**


*    Returns -1 for negative, 0 for zero, and 1 for positive numbers.

*  syntax : np.sign(arr)



###**1D Array**

```python
arr = np.array([-10,0,5,-3])
arr
```

```python
np.sign(arr)
```

###**2D Array**

```python
arr = np.array([[-10,0], [5,-3]])
arr
```

```python
np.sign(arr)
```
